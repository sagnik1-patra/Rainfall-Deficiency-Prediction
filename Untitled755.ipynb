```python
import os
import json
import yaml
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical


# ==============================
# Paths
# ==============================
DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Rainfall Deficiency Prediction\RJ_Session_247_AU_1955.2.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Rainfall Deficiency Prediction"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ==============================
# Load Dataset
# ==============================
df = pd.read_csv(DATASET_PATH)

print("Dataset Loaded Successfully")
print(df.head())
print(df.columns)


# ==============================
# Clean Dataset
# ==============================
df = df.dropna()

# Remove summary rows if needed
df = df[~df["State/Districts"].astype(str).str.contains("Rajasthan", case=False, na=False)]
df = df[~df["State/Districts"].astype(str).str.contains("East Rajasthan", case=False, na=False)]
df = df[~df["State/Districts"].astype(str).str.contains("West Rajasthan", case=False, na=False)]


# ==============================
# Create Target Column
# ==============================
def classify_rainfall(dep):
    if dep <= -20:
        return "Deficient"
    elif dep >= 20:
        return "Excess"
    else:
        return "Normal"


df["Rainfall_Status"] = df["% Dep."].apply(classify_rainfall)

df["Rainfall_Health_Score"] = df["Actual(mm)"] / df["Normal(mm)"]


# ==============================
# Encode Categorical Columns
# ==============================
region_encoder = LabelEncoder()
district_encoder = LabelEncoder()
target_encoder = LabelEncoder()

df["Region_Encoded"] = region_encoder.fit_transform(df["Region"])
df["District_Encoded"] = district_encoder.fit_transform(df["State/Districts"])
df["Target"] = target_encoder.fit_transform(df["Rainfall_Status"])


# ==============================
# Features and Target
# ==============================
X = df[
    [
        "Region_Encoded",
        "District_Encoded",
        "Actual(mm)",
        "Normal(mm)",
        "% Dep.",
        "Rainfall_Health_Score",
    ]
]

y = df["Target"]


# ==============================
# Scaling
# ==============================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# ==============================
# Train-Test Split
# ==============================
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y if len(y.unique()) > 1 else None
)


# ==============================
# Train Random Forest Model
# ==============================
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Random Forest Accuracy:", rf_accuracy)


# ==============================
# Save PKL Model
# ==============================
pkl_path = os.path.join(OUTPUT_DIR, "rainfall_deficiency_model.pkl")

joblib.dump(
    {
        "model": rf_model,
        "scaler": scaler,
        "region_encoder": region_encoder,
        "district_encoder": district_encoder,
        "target_encoder": target_encoder,
        "features": list(X.columns),
    },
    pkl_path
)

print("PKL model saved:", pkl_path)


# ==============================
# Train Deep Learning Model
# ==============================
num_classes = len(np.unique(y))

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

h5_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

h5_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = h5_model.fit(
    X_train,
    y_train_cat,
    epochs=50,
    batch_size=4,
    validation_data=(X_test, y_test_cat),
    verbose=1
)


# ==============================
# Save H5 Model
# ==============================
h5_path = os.path.join(OUTPUT_DIR, "rainfall_deficiency_model.h5")
h5_model.save(h5_path)

print("H5 model saved:", h5_path)


# ==============================
# Predictions
# ==============================
df["Predicted_Target"] = rf_model.predict(X_scaled)
df["Predicted_Status"] = target_encoder.inverse_transform(df["Predicted_Target"])

result_path = os.path.join(OUTPUT_DIR, "rainfall_prediction_results.csv")
df.to_csv(result_path, index=False)

print("Prediction CSV saved:", result_path)


# ==============================
# Save JSON Results
# ==============================
json_data = {
    "project_name": "Rainfall Deficiency Prediction",
    "model_used": "RandomForestClassifier + Neural Network",
    "accuracy": float(rf_accuracy),
    "features": list(X.columns),
    "target_classes": list(target_encoder.classes_),
    "total_records": int(len(df)),
    "output_files": {
        "pkl_model": pkl_path,
        "h5_model": h5_path,
        "csv_results": result_path
    }
}

json_path = os.path.join(OUTPUT_DIR, "rainfall_model_results.json")

with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)

print("JSON saved:", json_path)


# ==============================
# Save YAML Config
# ==============================
yaml_data = {
    "project": "Rainfall Deficiency Prediction",
    "dataset_path": DATASET_PATH,
    "output_dir": OUTPUT_DIR,
    "models": {
        "pkl": "RandomForestClassifier",
        "h5": "Keras Neural Network"
    },
    "features": list(X.columns),
    "target": "Rainfall_Status",
    "accuracy": float(rf_accuracy),
    "classes": list(target_encoder.classes_)
}

yaml_path = os.path.join(OUTPUT_DIR, "rainfall_model_config.yaml")

with open(yaml_path, "w") as f:
    yaml.dump(yaml_data, f, default_flow_style=False)

print("YAML saved:", yaml_path)


# ==============================
# Save Classification Report
# ==============================
report = classification_report(
    y_test,
    rf_predictions,
    target_names=target_encoder.classes_,
    zero_division=0
)

report_path = os.path.join(OUTPUT_DIR, "rainfall_classification_report.txt")

with open(report_path, "w") as f:
    f.write(report)

print("Classification report saved:", report_path)


print("\nAll files saved successfully!")
```
