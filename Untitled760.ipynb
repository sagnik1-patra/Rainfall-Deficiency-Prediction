```python
import os
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.ensemble import RandomForestClassifier

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical


# =========================================================
# PATHS
# =========================================================
DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Rainfall Deficiency Prediction\RJ_Session_247_AU_1955.2.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Rainfall Deficiency Prediction"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# =========================================================
# QPSO FEATURE SELECTION ALGORITHM
# Quantum Particle Swarm Optimization
# =========================================================
class QPSOFeatureSelector:
    def __init__(
        self,
        population_size=20,
        generations=20,
        alpha=0.75
    ):
        self.population_size = population_size
        self.generations = generations
        self.alpha = alpha
        self.best_solution = None
        self.best_score = 0
        self.history = []

    def fitness(self, solution, X_train, X_test, y_train, y_test):
        selected_features = np.where(solution == 1)[0]

        if len(selected_features) == 0:
            return 0

        model = RandomForestClassifier(
            n_estimators=100,
            random_state=42
        )

        model.fit(X_train[:, selected_features], y_train)
        predictions = model.predict(X_test[:, selected_features])

        return accuracy_score(y_test, predictions)

    def fit(self, X_train, X_test, y_train, y_test):
        n_features = X_train.shape[1]

        positions = np.random.randint(
            0,
            2,
            size=(self.population_size, n_features)
        )

        for i in range(self.population_size):
            if positions[i].sum() == 0:
                positions[i][np.random.randint(0, n_features)] = 1

        personal_best_positions = positions.copy()
        personal_best_scores = np.zeros(self.population_size)

        global_best_position = positions[0].copy()
        global_best_score = 0

        for generation in range(self.generations):
            for i in range(self.population_size):
                score = self.fitness(
                    positions[i],
                    X_train,
                    X_test,
                    y_train,
                    y_test
                )

                if score > personal_best_scores[i]:
                    personal_best_scores[i] = score
                    personal_best_positions[i] = positions[i].copy()

                if score > global_best_score:
                    global_best_score = score
                    global_best_position = positions[i].copy()

            mbest = np.mean(personal_best_positions, axis=0)

            for i in range(self.population_size):
                phi = np.random.rand(n_features)
                p = phi * personal_best_positions[i] + (1 - phi) * global_best_position

                u = np.random.rand(n_features)
                direction = np.where(np.random.rand(n_features) < 0.5, 1, -1)

                continuous_position = p + direction * self.alpha * np.abs(mbest - positions[i]) * np.log(1 / u)

                sigmoid = 1 / (1 + np.exp(-continuous_position))

                positions[i] = np.where(
                    np.random.rand(n_features) < sigmoid,
                    1,
                    0
                )

                if positions[i].sum() == 0:
                    positions[i][np.random.randint(0, n_features)] = 1

            self.history.append(global_best_score)

            print(
                f"Generation {generation + 1}/{self.generations} "
                f"| Best QPSO Accuracy: {global_best_score:.4f}"
            )

        self.best_solution = global_best_position
        self.best_score = global_best_score

        return self.best_solution


# =========================================================
# LOAD DATASET
# =========================================================
df = pd.read_csv(DATASET_PATH)
df.columns = df.columns.str.strip()

print("Dataset Loaded Successfully")
print(df.head())
print(df.columns)


# =========================================================
# CLEAN DATASET
# =========================================================
df = df.dropna()

summary_rows = [
    "Rajasthan",
    "East Rajasthan",
    "West Rajasthan"
]

for row in summary_rows:
    df = df[
        ~df["State/Districts"]
        .astype(str)
        .str.contains(row, case=False, na=False)
    ]


# =========================================================
# TARGET CREATION
# =========================================================
def classify_rainfall(dep):
    if dep <= -20:
        return "Deficient"
    elif dep >= 20:
        return "Excess"
    else:
        return "Normal"


df["Rainfall_Status"] = df["% Dep."].apply(classify_rainfall)

df["Rainfall_Health_Score"] = (
    df["Actual(mm)"] / df["Normal(mm)"]
)


# =========================================================
# LABEL ENCODING
# =========================================================
region_encoder = LabelEncoder()
district_encoder = LabelEncoder()
target_encoder = LabelEncoder()

df["Region_Encoded"] = region_encoder.fit_transform(df["Region"])
df["District_Encoded"] = district_encoder.fit_transform(df["State/Districts"])
df["Target"] = target_encoder.fit_transform(df["Rainfall_Status"])


# =========================================================
# FEATURES
# =========================================================
all_features = [
    "Region_Encoded",
    "District_Encoded",
    "Actual(mm)",
    "Normal(mm)",
    "% Dep.",
    "Rainfall_Health_Score"
]

X = df[all_features]
y = df["Target"]


# =========================================================
# FEATURE SCALING
# =========================================================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# =========================================================
# TRAIN TEST SPLIT
# =========================================================
stratify_value = y if len(y.unique()) > 1 else None

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=stratify_value
)


# =========================================================
# QPSO FEATURE SELECTION
# =========================================================
qpso = QPSOFeatureSelector(
    population_size=20,
    generations=20,
    alpha=0.75
)

best_solution = qpso.fit(
    X_train,
    X_test,
    y_train,
    y_test
)

selected_feature_indices = np.where(best_solution == 1)[0]

selected_features = [
    all_features[i]
    for i in selected_feature_indices
]

print("\nQPSO Selected Features:")
print(selected_features)


# =========================================================
# SELECT QPSO FEATURES
# =========================================================
X_train_selected = X_train[:, selected_feature_indices]
X_test_selected = X_test[:, selected_feature_indices]
X_scaled_selected = X_scaled[:, selected_feature_indices]


# =========================================================
# FINAL QPSO RANDOM FOREST MODEL
# =========================================================
qpso_model = RandomForestClassifier(
    n_estimators=150,
    random_state=42
)

qpso_model.fit(X_train_selected, y_train)

y_pred = qpso_model.predict(X_test_selected)

qpso_accuracy = accuracy_score(y_test, y_pred)

print("\nFinal QPSO Accuracy:", qpso_accuracy)


# =========================================================
# DEEP LEARNING MODEL
# =========================================================
num_classes = len(np.unique(y))

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

h5_model = Sequential([
    Dense(
        64,
        activation="relu",
        input_shape=(X_train_selected.shape[1],)
    ),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

h5_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = h5_model.fit(
    X_train_selected,
    y_train_cat,
    epochs=50,
    batch_size=4,
    validation_data=(X_test_selected, y_test_cat),
    verbose=1
)


# =========================================================
# SAVE MODELS
# =========================================================
pkl_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_deficiency_model.pkl"
)

h5_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_deficiency_model.h5"
)

joblib.dump(
    {
        "model": qpso_model,
        "scaler": scaler,
        "selected_features": selected_features,
        "selected_feature_indices": selected_feature_indices.tolist(),
        "region_encoder": region_encoder,
        "district_encoder": district_encoder,
        "target_encoder": target_encoder,
        "all_features": all_features,
        "algorithm": "Quantum Particle Swarm Optimization"
    },
    pkl_path
)

h5_model.save(h5_path)


# =========================================================
# PREDICTIONS
# =========================================================
df["Predicted_Target"] = qpso_model.predict(X_scaled_selected)

df["Predicted_Status"] = target_encoder.inverse_transform(
    df["Predicted_Target"]
)

df["Actual_Status"] = df["Rainfall_Status"]


# =========================================================
# SAVE RESULT CSV
# =========================================================
result_csv_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_result.csv"
)

df.to_csv(result_csv_path, index=False)


# =========================================================
# SAVE PREDICTION CSV
# =========================================================
prediction_df = df[
    [
        "Region",
        "State/Districts",
        "Actual(mm)",
        "Normal(mm)",
        "% Dep.",
        "Rainfall_Health_Score",
        "Actual_Status",
        "Predicted_Status"
    ]
]

prediction_csv_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_prediction.csv"
)

prediction_df.to_csv(prediction_csv_path, index=False)


# =========================================================
# SAVE JSON
# =========================================================
json_data = {
    "project_name": "Rainfall Deficiency Prediction using QPSO",
    "algorithm": "Quantum Particle Swarm Optimization",
    "accuracy": float(qpso_accuracy),
    "selected_features": selected_features,
    "all_features": all_features,
    "classes": list(target_encoder.classes_),
    "total_records": int(len(df)),
    "files": {
        "pkl": pkl_path,
        "h5": h5_path,
        "result_csv": result_csv_path,
        "prediction_csv": prediction_csv_path
    }
}

json_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_model_results.json"
)

with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)


# =========================================================
# SAVE YAML
# =========================================================
yaml_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_model_config.yaml"
)

with open(yaml_path, "w") as f:
    yaml.dump(json_data, f, default_flow_style=False)


# =========================================================
# CLASSIFICATION REPORT
# =========================================================
report = classification_report(
    y_test,
    y_pred,
    target_names=target_encoder.classes_,
    zero_division=0
)

report_path = os.path.join(
    OUTPUT_DIR,
    "qpso_rainfall_classification_report.txt"
)

with open(report_path, "w") as f:
    f.write(report)


# =========================================================
# QPSO ACCURACY GRAPH
# =========================================================
plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(qpso.history) + 1),
    qpso.history,
    marker="o"
)

plt.title("QPSO Optimization Accuracy Graph")
plt.xlabel("Generation")
plt.ylabel("Best Accuracy")
plt.grid(True)

accuracy_graph_path = os.path.join(
    OUTPUT_DIR,
    "qpso_accuracy_graph.png"
)

plt.savefig(
    accuracy_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# NEURAL NETWORK ACCURACY GRAPH
# =========================================================
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("QPSO Neural Network Accuracy Graph")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

nn_graph_path = os.path.join(
    OUTPUT_DIR,
    "qpso_nn_accuracy_graph.png"
)

plt.savefig(
    nn_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# PREDICTION GRAPH
# =========================================================
prediction_counts = prediction_df[
    "Predicted_Status"
].value_counts()

plt.figure(figsize=(8, 5))
prediction_counts.plot(kind="bar")

plt.title("QPSO Prediction Graph")
plt.xlabel("Predicted Status")
plt.ylabel("District Count")
plt.grid(axis="y")

prediction_graph_path = os.path.join(
    OUTPUT_DIR,
    "qpso_prediction_graph.png"
)

plt.savefig(
    prediction_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# RESULT GRAPH
# =========================================================
actual_counts = df[
    "Actual_Status"
].value_counts()

plt.figure(figsize=(8, 5))
actual_counts.plot(kind="bar")

plt.title("QPSO Result Graph")
plt.xlabel("Actual Status")
plt.ylabel("District Count")
plt.grid(axis="y")

result_graph_path = os.path.join(
    OUTPUT_DIR,
    "qpso_result_graph.png"
)

plt.savefig(
    result_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# COMPARISON GRAPH
# =========================================================
comparison_df = pd.DataFrame({
    "Actual": df["Actual_Status"].value_counts(),
    "Predicted": df["Predicted_Status"].value_counts()
}).fillna(0)

comparison_df.plot(
    kind="bar",
    figsize=(9, 6)
)

plt.title("QPSO Actual vs Predicted Comparison")
plt.xlabel("Rainfall Status")
plt.ylabel("District Count")
plt.grid(axis="y")

comparison_graph_path = os.path.join(
    OUTPUT_DIR,
    "qpso_comparison_graph.png"
)

plt.savefig(
    comparison_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# HEATMAP
# =========================================================
plt.figure(figsize=(10, 7))

heatmap_cols = [
    "Actual(mm)",
    "Normal(mm)",
    "% Dep.",
    "Rainfall_Health_Score",
    "Region_Encoded",
    "District_Encoded",
    "Target",
    "Predicted_Target"
]

corr = df[heatmap_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("QPSO Heatmap")

heatmap_path = os.path.join(
    OUTPUT_DIR,
    "qpso_heatmap.png"
)

plt.savefig(
    heatmap_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# CONFUSION MATRIX
# =========================================================
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=target_encoder.classes_,
    yticklabels=target_encoder.classes_
)

plt.title("QPSO Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

confusion_matrix_path = os.path.join(
    OUTPUT_DIR,
    "qpso_confusion_matrix.png"
)

plt.savefig(
    confusion_matrix_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# DISTRICT COMPARISON GRAPH
# =========================================================
plt.figure(figsize=(14, 7))

plt.plot(
    df["State/Districts"],
    df["Actual(mm)"],
    marker="o",
    label="Actual Rainfall"
)

plt.plot(
    df["State/Districts"],
    df["Normal(mm)"],
    marker="x",
    label="Normal Rainfall"
)

plt.title("QPSO District Rainfall Comparison")
plt.xlabel("District")
plt.ylabel("Rainfall (mm)")
plt.xticks(rotation=90)
plt.legend()
plt.grid(True)

district_graph_path = os.path.join(
    OUTPUT_DIR,
    "qpso_district_comparison_graph.png"
)

plt.savefig(
    district_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =========================================================
# FINAL OUTPUT
# =========================================================
print("\nAll QPSO files saved successfully!")

print("\nSaved Files:")
print("PKL Model:", pkl_path)
print("H5 Model:", h5_path)
print("Result CSV:", result_csv_path)
print("Prediction CSV:", prediction_csv_path)
print("JSON:", json_path)
print("YAML:", yaml_path)
print("Classification Report:", report_path)
print("Accuracy Graph:", accuracy_graph_path)
print("NN Accuracy Graph:", nn_graph_path)
print("Prediction Graph:", prediction_graph_path)
print("Result Graph:", result_graph_path)
print("Comparison Graph:", comparison_graph_path)
print("Heatmap:", heatmap_path)
print("Confusion Matrix:", confusion_matrix_path)
print("District Graph:", district_graph_path)
```
