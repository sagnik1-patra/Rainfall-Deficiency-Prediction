```python
import os
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical


# ==============================
# Paths
# ==============================
DATASET_PATH = r"C:\Users\NXTWAVE\Downloads\Rainfall Deficiency Prediction\RJ_Session_247_AU_1955.2.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Rainfall Deficiency Prediction"

os.makedirs(OUTPUT_DIR, exist_ok=True)


# ==============================
# Load Dataset
# ==============================
df = pd.read_csv(DATASET_PATH)

print("Dataset Loaded Successfully")
print(df.head())
print(df.columns)


# ==============================
# Clean Column Names
# ==============================
df.columns = df.columns.str.strip()

# Adjust column names if needed
print("Available Columns:", df.columns.tolist())


# ==============================
# Clean Dataset
# ==============================
df = df.dropna()

summary_words = ["Rajasthan", "East Rajasthan", "West Rajasthan"]
for word in summary_words:
    df = df[~df["State/Districts"].astype(str).str.contains(word, case=False, na=False)]


# ==============================
# Create Target Column
# ==============================
def classify_rainfall(dep):
    if dep <= -20:
        return "Deficient"
    elif dep >= 20:
        return "Excess"
    else:
        return "Normal"


df["Rainfall_Status"] = df["% Dep."].apply(classify_rainfall)
df["Rainfall_Health_Score"] = df["Actual(mm)"] / df["Normal(mm)"]


# ==============================
# Encode Columns
# ==============================
region_encoder = LabelEncoder()
district_encoder = LabelEncoder()
target_encoder = LabelEncoder()

df["Region_Encoded"] = region_encoder.fit_transform(df["Region"])
df["District_Encoded"] = district_encoder.fit_transform(df["State/Districts"])
df["Target"] = target_encoder.fit_transform(df["Rainfall_Status"])


# ==============================
# Features and Target
# ==============================
features = [
    "Region_Encoded",
    "District_Encoded",
    "Actual(mm)",
    "Normal(mm)",
    "% Dep.",
    "Rainfall_Health_Score"
]

X = df[features]
y = df["Target"]


# ==============================
# Scaling
# ==============================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# ==============================
# Train-Test Split
# ==============================
stratify_value = y if len(y.unique()) > 1 else None

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=stratify_value
)


# ==============================
# Random Forest Model
# ==============================
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Random Forest Accuracy:", rf_accuracy)


# ==============================
# Deep Learning Model
# ==============================
num_classes = len(np.unique(y))

y_train_cat = to_categorical(y_train, num_classes=num_classes)
y_test_cat = to_categorical(y_test, num_classes=num_classes)

h5_model = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.2),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(num_classes, activation="softmax")
])

h5_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history = h5_model.fit(
    X_train,
    y_train_cat,
    epochs=50,
    batch_size=4,
    validation_data=(X_test, y_test_cat),
    verbose=1
)


# ==============================
# Save Models
# ==============================
pkl_path = os.path.join(OUTPUT_DIR, "rainfall_deficiency_model.pkl")
h5_path = os.path.join(OUTPUT_DIR, "rainfall_deficiency_model.h5")

joblib.dump(
    {
        "model": rf_model,
        "scaler": scaler,
        "region_encoder": region_encoder,
        "district_encoder": district_encoder,
        "target_encoder": target_encoder,
        "features": features,
    },
    pkl_path
)

h5_model.save(h5_path)


# ==============================
# Full Dataset Predictions
# ==============================
df["Predicted_Target"] = rf_model.predict(X_scaled)
df["Predicted_Status"] = target_encoder.inverse_transform(df["Predicted_Target"])

df["Actual_Status"] = df["Rainfall_Status"]


# ==============================
# Save Result CSV
# ==============================
result_csv_path = os.path.join(OUTPUT_DIR, "rainfall_result.csv")
df.to_csv(result_csv_path, index=False)


# ==============================
# Save Prediction CSV
# ==============================
prediction_df = df[
    [
        "Region",
        "State/Districts",
        "Actual(mm)",
        "Normal(mm)",
        "% Dep.",
        "Rainfall_Health_Score",
        "Actual_Status",
        "Predicted_Status"
    ]
]

prediction_csv_path = os.path.join(OUTPUT_DIR, "rainfall_prediction.csv")
prediction_df.to_csv(prediction_csv_path, index=False)


# ==============================
# JSON and YAML
# ==============================
json_data = {
    "project_name": "Rainfall Deficiency Prediction",
    "accuracy": float(rf_accuracy),
    "features": features,
    "classes": list(target_encoder.classes_),
    "total_records": int(len(df)),
    "files": {
        "pkl": pkl_path,
        "h5": h5_path,
        "result_csv": result_csv_path,
        "prediction_csv": prediction_csv_path
    }
}

json_path = os.path.join(OUTPUT_DIR, "rainfall_model_results.json")

with open(json_path, "w") as f:
    json.dump(json_data, f, indent=4)


yaml_path = os.path.join(OUTPUT_DIR, "rainfall_model_config.yaml")

with open(yaml_path, "w") as f:
    yaml.dump(json_data, f, default_flow_style=False)


# ==============================
# Classification Report
# ==============================
report = classification_report(
    y_test,
    rf_predictions,
    target_names=target_encoder.classes_,
    zero_division=0
)

report_path = os.path.join(OUTPUT_DIR, "rainfall_classification_report.txt")

with open(report_path, "w") as f:
    f.write(report)


# ==============================
# 1. Accuracy Graph
# ==============================
plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Model Accuracy Graph")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

accuracy_graph_path = os.path.join(OUTPUT_DIR, "rainfall_accuracy_graph.png")
plt.savefig(accuracy_graph_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# 2. Prediction Graph
# ==============================
prediction_counts = prediction_df["Predicted_Status"].value_counts()

plt.figure(figsize=(8, 5))
prediction_counts.plot(kind="bar")
plt.title("Predicted Rainfall Status Graph")
plt.xlabel("Predicted Status")
plt.ylabel("Number of Districts")
plt.grid(axis="y")

prediction_graph_path = os.path.join(OUTPUT_DIR, "rainfall_prediction_graph.png")
plt.savefig(prediction_graph_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# 3. Result Graph
# ==============================
actual_counts = df["Actual_Status"].value_counts()

plt.figure(figsize=(8, 5))
actual_counts.plot(kind="bar")
plt.title("Actual Rainfall Status Result Graph")
plt.xlabel("Actual Status")
plt.ylabel("Number of Districts")
plt.grid(axis="y")

result_graph_path = os.path.join(OUTPUT_DIR, "rainfall_result_graph.png")
plt.savefig(result_graph_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# 4. Comparison Graph
# ==============================
comparison_df = pd.DataFrame({
    "Actual": df["Actual_Status"].value_counts(),
    "Predicted": df["Predicted_Status"].value_counts()
}).fillna(0)

comparison_df.plot(kind="bar", figsize=(9, 6))
plt.title("Actual vs Predicted Rainfall Status Comparison")
plt.xlabel("Rainfall Status")
plt.ylabel("Number of Districts")
plt.grid(axis="y")

comparison_graph_path = os.path.join(OUTPUT_DIR, "rainfall_comparison_graph.png")
plt.savefig(comparison_graph_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# 5. Heatmap
# ==============================
plt.figure(figsize=(10, 7))
corr = df[
    [
        "Actual(mm)",
        "Normal(mm)",
        "% Dep.",
        "Rainfall_Health_Score",
        "Region_Encoded",
        "District_Encoded",
        "Target",
        "Predicted_Target"
    ]
].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Rainfall Feature Correlation Heatmap")

heatmap_path = os.path.join(OUTPUT_DIR, "rainfall_heatmap.png")
plt.savefig(heatmap_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# 6. Confusion Matrix Graph
# ==============================
cm = confusion_matrix(y_test, rf_predictions)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=target_encoder.classes_,
    yticklabels=target_encoder.classes_
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

confusion_matrix_path = os.path.join(OUTPUT_DIR, "rainfall_confusion_matrix.png")
plt.savefig(confusion_matrix_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# 7. District-wise Rainfall Comparison
# ==============================
plt.figure(figsize=(14, 7))
plt.plot(df["State/Districts"], df["Actual(mm)"], marker="o", label="Actual Rainfall")
plt.plot(df["State/Districts"], df["Normal(mm)"], marker="x", label="Normal Rainfall")
plt.title("District-wise Actual vs Normal Rainfall")
plt.xlabel("District")
plt.ylabel("Rainfall in mm")
plt.xticks(rotation=90)
plt.legend()
plt.grid(True)

district_comparison_path = os.path.join(OUTPUT_DIR, "rainfall_district_comparison_graph.png")
plt.savefig(district_comparison_path, dpi=300, bbox_inches="tight")
plt.show()


# ==============================
# Final Output
# ==============================
print("\nAll files saved successfully!")

print("\nSaved Files:")
print("H5 Model:", h5_path)
print("PKL Model:", pkl_path)
print("YAML File:", yaml_path)
print("JSON File:", json_path)
print("Result CSV:", result_csv_path)
print("Prediction CSV:", prediction_csv_path)
print("Accuracy Graph:", accuracy_graph_path)
print("Prediction Graph:", prediction_graph_path)
print("Result Graph:", result_graph_path)
print("Comparison Graph:", comparison_graph_path)
print("Heatmap:", heatmap_path)
print("Confusion Matrix:", confusion_matrix_path)
print("District Comparison Graph:", district_comparison_path)
```
